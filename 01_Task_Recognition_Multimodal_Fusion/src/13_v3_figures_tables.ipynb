# v3 rebuild — publication figures & LaTeX tables

Reads the sealed results in `v3_rebuild/benchmark/` and `v3_rebuild/ablation/`
(including the corrected two-tower), regenerates the six figures and the four LaTeX
tables, and copies the figures into `paper/figures/`. Re-running this notebook is
cheap (seconds); it does not re-run any experiment.


```python
import sys, shutil
from pathlib import Path

def _root(start=Path.cwd()):
    for d in [start, *start.parents]:
        if (d / "src" / "rebuild" / "models_registry.py").exists():
            return d
    raise SystemExit("run notebook from the module root or src/")
REPO = _root()
sys.path.insert(0, str(REPO / "src" / "rebuild"))

import make_figures as F
import make_latex_tables as T

print("figure dir :", F.FIG)
print("table dir  :", T.TAB)
```

### 6 figures

```python
F.fig_confusion(); F.fig_perclass(); F.fig_subject_box()
F.fig_ablation(); F.fig_gap(); F.fig_significance()
```

### 4 LaTeX tables

```python
T.tbl_benchmark(); T.tbl_perclass(); T.tbl_ablation(); T.tbl_gap()
```

### Copy figures into paper/figures

```python
FIGDST = REPO / "paper" / "figures"
FIGDST.mkdir(parents=True, exist_ok=True)
for p in sorted(F.FIG.glob("fig*.png")):
    shutil.copy(p, FIGDST / p.name)
print("figures in paper/figures:")
for p in sorted(FIGDST.glob("*.png")):
    print("   ", p.name)
```
